```python
-- 1. Create Database

CREATE DATABASE workforce_analytics;

USE workforce_analytics;


-- 2. Create Employee Table


CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    employee_name VARCHAR(100),
    department VARCHAR(50),
    job_role VARCHAR(100),
    age INT,
    gender VARCHAR(20),
    years_at_company INT,
    monthly_income DECIMAL(10,2),
    performance_score DECIMAL(5,2),
    job_satisfaction DECIMAL(5,2),
    training_hours INT,
    overtime VARCHAR(10),
    promotion_years INT,
    attrition VARCHAR(10)
);


-- 3. Insert Sample Workforce Data

INSERT INTO employees VALUES
(101, 'Employee A', 'IT', 'Data Analyst', 25, 'Female',
 2, 45000, 82, 4, 40, 'No', 1, 'No'),

(102, 'Employee B', 'IT', 'Software Engineer', 29, 'Male',
 5, 65000, 76, 2, 20, 'Yes', 3, 'Yes'),

(103, 'Employee C', 'Sales', 'Sales Executive', 31, 'Female',
 4, 38000, 72, 2, 15, 'Yes', 4, 'Yes'),

(104, 'Employee D', 'HR', 'HR Executive', 28, 'Female',
 3, 42000, 88, 5, 50, 'No', 1, 'No'),

(105, 'Employee E', 'Finance', 'Financial Analyst', 35, 'Male',
 7, 72000, 91, 4, 60, 'No', 2, 'No'),

(106, 'Employee F', 'Sales', 'Sales Manager', 40, 'Male',
 9, 58000, 69, 2, 10, 'Yes', 5, 'Yes'),

(107, 'Employee G', 'IT', 'ML Engineer', 27, 'Female',
 3, 70000, 94, 5, 80, 'No', 1, 'No'),

(108, 'Employee H', 'Marketing', 'Marketing Analyst', 30, 'Female',
 4, 47000, 79, 3, 35, 'No', 3, 'No');



-- 4. View Complete Employee Data


SELECT *
FROM employees;



-- 5. Total Number of Employees


SELECT COUNT(*) AS total_employees
FROM employees;



-- 6. Attrition Analysis


SELECT
    attrition,
    COUNT(*) AS employee_count
FROM employees
GROUP BY attrition;



-- 7. Overall Attrition Rate


SELECT
    ROUND(
        SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END)
        * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees;



-- 8. Department-wise Employee Count


SELECT
    department,
    COUNT(*) AS employee_count
FROM employees
GROUP BY department
ORDER BY employee_count DESC;



-- 9. Department-wise Attrition


SELECT
    department,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END)
        AS employees_left,
    ROUND(
        SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END)
        * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY department
ORDER BY attrition_rate DESC;



-- 10. Average Performance by Department


SELECT
    department,
    ROUND(AVG(performance_score), 2)
        AS average_performance
FROM employees
GROUP BY department
ORDER BY average_performance DESC;



-- 11. Average Job Satisfaction by Department


SELECT
    department,
    ROUND(AVG(job_satisfaction), 2)
        AS average_satisfaction
FROM employees
GROUP BY department
ORDER BY average_satisfaction DESC;



-- 12. Training Analysis


SELECT
    department,
    ROUND(AVG(training_hours), 2)
        AS average_training_hours
FROM employees
GROUP BY department
ORDER BY average_training_hours DESC;



-- 13. Salary Analysis


SELECT
    department,
    ROUND(AVG(monthly_income), 2)
        AS average_monthly_income
FROM employees
GROUP BY department
ORDER BY average_monthly_income DESC;



-- 14. High Performing Employees


SELECT
    employee_id,
    employee_name,
    department,
    job_role,
    performance_score
FROM employees
WHERE performance_score >= 85
ORDER BY performance_score DESC;



-- 15. Employees with Low Job Satisfaction


SELECT
    employee_id,
    employee_name,
    department,
    job_role,
    job_satisfaction
FROM employees
WHERE job_satisfaction <= 2
ORDER BY job_satisfaction ASC;



-- 16. Employees Working Overtime


SELECT
    employee_id,
    employee_name,
    department,
    job_role,
    overtime
FROM employees
WHERE overtime = 'Yes';



-- 17. Potential High-Risk Employees


SELECT
    employee_id,
    employee_name,
    department,
    job_role,
    job_satisfaction,
    overtime,
    promotion_years,
    attrition
FROM employees
WHERE job_satisfaction <= 2
   OR overtime = 'Yes'
   OR promotion_years >= 4;



-- 18. Employees Requiring Career Development


SELECT
    employee_id,
    employee_name,
    department,
    job_role,
    performance_score,
    training_hours
FROM employees
WHERE training_hours < 30
AND performance_score >= 70
ORDER BY performance_score DESC;



-- 19. Department Workforce Health


SELECT
    department,
    COUNT(*) AS total_employees,
    ROUND(AVG(performance_score), 2)
        AS avg_performance,
    ROUND(AVG(job_satisfaction), 2)
        AS avg_satisfaction,
    ROUND(AVG(training_hours), 2)
        AS avg_training_hours,
    ROUND(
        SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END)
        * 100.0 / COUNT(*),
        2
    ) AS attrition_rate
FROM employees
GROUP BY department
ORDER BY attrition_rate DESC;



-- 20. Workforce Risk Classification


SELECT
    employee_id,
    employee_name,
    department,
    job_role,

    CASE
        WHEN job_satisfaction <= 2
             AND overtime = 'Yes'
             AND promotion_years >= 4
            THEN 'High Risk'

        WHEN job_satisfaction <= 3
             OR overtime = 'Yes'
             OR promotion_years >= 4
            THEN 'Medium Risk'

        ELSE 'Low Risk'
    END AS workforce_risk

FROM employees;



-- 21. Top Performing Departments


SELECT
    department,
    ROUND(AVG(performance_score), 2)
        AS average_performance
FROM employees
GROUP BY department
HAVING AVG(performance_score) >= 80
ORDER BY average_performance DESC;



-- 22. Workforce Summary


SELECT

    COUNT(*) AS total_employees,

    SUM(CASE
        WHEN attrition = 'Yes' THEN 1
        ELSE 0
    END) AS employees_left,

    SUM(CASE
        WHEN attrition = 'No' THEN 1
        ELSE 0
    END) AS employees_retained,

    ROUND(AVG(performance_score), 2)
        AS average_performance,

    ROUND(AVG(job_satisfaction), 2)
        AS average_satisfaction,

    ROUND(AVG(training_hours), 2)
        AS average_training_hours,

    ROUND(AVG(monthly_income), 2)
        AS average_income,

    ROUND(
        SUM(CASE WHEN attrition = 'Yes' THEN 1 ELSE 0 END)
        * 100.0 / COUNT(*),
        2
    ) AS attrition_rate

FROM employees;
```
