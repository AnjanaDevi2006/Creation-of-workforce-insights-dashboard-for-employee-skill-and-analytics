```python
import csv
import json
import os
```

```python
def analyze_workforce(input_file, output_file):

    # Initialize workforce statistics
    report = {
        "total_employees": 0,
        "employees_left": 0,
        "employees_stayed": 0,
        "average_performance": 0,
        "average_job_satisfaction": 0,
        "average_training_hours": 0,
        "department_employee_count": {}
    }

    # Handle missing file
    if not os.path.exists(input_file):
        print(f"Error: The file '{input_file}' was not found.")
        return

    try:
        performance_total = 0
        satisfaction_total = 0
        training_total = 0

        # Read employee dataset
        with open(input_file, "r", encoding="utf-8") as file:

            reader = csv.DictReader(file)

            for row in reader:

                report["total_employees"] += 1

                # Count attrition
                if row["Attrition"].strip().lower() == "yes":
                    report["employees_left"] += 1
                else:
                    report["employees_stayed"] += 1

                # Calculate workforce metrics
                performance_total += float(row["PerformanceScore"])
                satisfaction_total += float(row["JobSatisfaction"])
                training_total += float(row["TrainingHours"])

                # Department-wise employee count
                department = row["Department"]

                if department not in report["department_employee_count"]:
                    report["department_employee_count"][department] = 0

                report["department_employee_count"][department] += 1

        # Calculate averages
        if report["total_employees"] > 0:

            report["average_performance"] = round(
                performance_total / report["total_employees"], 2
            )

            report["average_job_satisfaction"] = round(
                satisfaction_total / report["total_employees"], 2
            )

            report["average_training_hours"] = round(
                training_total / report["total_employees"], 2
            )

        # Calculate attrition rate
        if report["total_employees"] > 0:
            report["attrition_rate"] = round(
                (report["employees_left"] /
                 report["total_employees"]) * 100,
                2
            )
        else:
            report["attrition_rate"] = 0

        # Print results
        print("\n--- Workforce Analytics Summary ---")

        print(f"Total Employees: {report['total_employees']}")
        print(f"Employees Left: {report['employees_left']}")
        print(f"Employees Stayed: {report['employees_stayed']}")
        print(f"Attrition Rate: {report['attrition_rate']}%")
        print(f"Average Performance: {report['average_performance']}")
        print(
            f"Average Job Satisfaction: "
            f"{report['average_job_satisfaction']}"
        )
        print(
            f"Average Training Hours: "
            f"{report['average_training_hours']}"
        )

        print("\nDepartment-wise Employee Count:")

        for department, count in report[
            "department_employee_count"
        ].items():

            print(f"{department}: {count}")

        # Save results into JSON
        with open(output_file, "w", encoding="utf-8") as json_file:

            json.dump(
                report,
                json_file,
                indent=4
            )

        print(f"\nWorkforce summary successfully saved to {output_file}")

    except (ValueError, KeyError) as e:

        print(f"Data processing error: {e}")

    except Exception as e:

        print(f"An unexpected error occurred: {e}")


if __name__ == "__main__":

    # Input and output files
    employee_file = "employee_data.csv"
    summary_file = "workforce_summary.json"

    # Run workforce analysis
    analyze_workforce(
        employee_file,
        summary_file
    )

```
